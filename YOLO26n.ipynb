### 🔎 Using Ultralytics YOLO26n for Object Detection  
* Please run under a virtual environment (e.g. yolo26-env) to isolate conflicting version  
* Requires `Python 3.11` and `ultralytics` installation

```python
from ultralytics import YOLO

# Filters existing COCO dataset for outside objects
# person, bicycle, car, motorcycle, bus, truck, traffic light, fire hydrant, stop sign, bench, dog
OUTDOOR = [0, 1, 2, 3, 5, 7, 9, 10, 11, 13, 16]

# Load pretrained YOLO26n model using COCO weights
model = YOLO("yolo26n.pt")

# Rename all labels as "object"
model.model.names = {i: "object" for i in range(len(model.names))}

# model.save("yolo26n_object.pt") # Export to Orin

# Run inference
results = model.predict("https://ultralytics.com/images/bus.jpg", classes=OUTDOOR, save=True, conf=0.25)
```
